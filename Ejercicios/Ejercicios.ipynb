## 1.3 Valor de estrategias mixtas y equilibrio de Nash

<p>Calcule el valor de la estrategia mixta 
\begin{align*}
    \pi_1 &= \left (\frac{1}{4}, \frac{3}{4}, 0 \right)
    &
    \pi_2 &= \left (\frac{1}{2}, \frac{1}{2} \right)
\end{align*}

\begin{align*}
V_1(\pi) &= \frac{1}{2} \left( \frac{1}{4} 0 + \frac{3}{4} 4 + 0 \cdot 3 \right) 
          + \frac{1}{2} \left( \frac{1}{4} 6 + \frac{3}{4} 2 + 0 \cdot 4 \right)
          = \frac{1}{8} 6 + \frac{3}{8} 6 = 3 \\
V_2(\pi) &= \frac{1}{4} \left( \frac{1}{2} 1 + \frac{1}{2} 5 \right) 
          + \frac{3}{4} \left( \frac{1}{2} 4 + \frac{1}{2} 0 \right) 
          + 0 \left( \frac{1}{2} 0 + \frac{1}{2} 2 \right) 
          = \frac{3}{4} + \frac{3}{2} = \frac{9}{4}
\end{align*}

<p>¿Es un equilibrio de Nash (en estrategias mixtas)? No, porque $P1$ puede cambiar a la estrategia pura $C = (0, 0, 1)$ y obtener un mayor valor:

\begin{align*}
V_1(\pi) &= \frac{1}{2} \left( 0 \cdot 0 + 0 \cdot 4 + 1 \cdot 3 \right) 
          + \frac{1}{2} \left( 0 \cdot 6 + 0 \cdot 2 + 1 \cdot 4 \right)
          = \frac{1}{2} 3 + \frac{1}{2} 4 =  \frac{7}{2} > 3 \\
\end{align*}

## 1.4 Equilibrios de Nash en estrategias puras y mixtas

```python
A_Q1 = np.array(['A', 'B'])
A_Q2 = np.array(['D', 'E'])
R_Q1 = np.array([[2, 8], [6, 4]])
R_Q2 = np.array([[3, 6], [5, 2]])
G_1_4 = { 'Q1' : (R_Q1, A_Q1), 'Q2' : (R_Q2, A_Q2) }
```

```python
nash(G_1_4)
```

<p>Sea $\pi = (\pi_1, \pi_2)$ la estrategia mixta
   \begin{gather*}
    \pi_1 = \bigg( \frac{2}{3}, \frac{1}{3} \bigg) &
    \pi_2 = \bigg( \frac{1}{2}, \frac{1}{2} \bigg)
   \end{gather*}

<p>Verificamos que $(\pi_1, \pi_2)$ es NE usando la definición de NE: $\pi_i \in BR(\pi_j), i \neq j$
<p>Sea $\pi'_1 = (p, 1-p)$
   \begin{align*}
    V_1(\pi'_1, \pi_2) &= 
    \frac{1}{2} \bigg( 2 p  + 6 (1 - p) \bigg) + \frac{1}{2} \bigg( 8 p + 4 (1 - p) \bigg) 
    = p + 3 (1 - p) + 4 p + 2 (1 - p) 
    = 5 \\
    &\implies \forall \pi'_1.\ \pi'_1 \in BR(\pi_2) \implies \pi_1 \in BR(\pi_2)
   \end{align*}
<p>Sea $\pi'_2 = (q, 1-q)$
   \begin{align*}
    V_2(\pi_1, \pi'_2) &= 
    \frac{2}{3} \bigg( 3 q  + 5 (1 - q) \bigg) + \frac{1}{3} \bigg( 6 q + 2 (1 - q) \bigg) 
    = 2 q + \frac{10}{3} (1 - q) + 2 q + \frac{2}{3} (1 - q) = 4 \\
    &\implies \forall \pi'_2.\ \pi'_2 \in BR(\pi_1) \implies \pi_2 \in BR(\pi_1)
   \end{align*}

```python
X = np.arange(0, 1, 0.01)
```

<p>Max-Min: $\max_p\min\{2p-1, -3p+1\}$

```python
V_L = np.array([ 2*p-1 for p in X ])
V_R = np.array([ -3*p+1 for p in X ])
plt.plot(X, V_L, label = 'V_L')
plt.plot(X, V_R, label = 'V_R')
plt.axhline(y = -0.2, linestyle = 'dashed', label = 'V Max-Min')
plt.axvline(x = 0.4, linestyle = 'dotted', label = 'p*')
plt.xlabel('p')
plt.ylabel('V')
plt.title('Max-Min')
plt.legend()
plt.show()
```

<p>Min-Max: $\min_q\max\{3q-2, -2q+1\}$

```python
V_U = np.array([ 3*q-2 for q in X ])
V_D = np.array([ -2*q+1 for q in X ])
plt.plot(X, V_U, label = 'V_U')
plt.plot(X, V_D, label = 'V_D')
plt.axhline(y = -0.2, linestyle = 'dashed', label = 'V Min-Max')
plt.axvline(x = 0.6, linestyle = 'dotted', label = 'q*')
plt.xlabel('q')
plt.ylabel('V')
plt.title('Min-Max')
plt.legend()
plt.show()
```
