```python
A_A = np.array(['a1', 'a2'])
A_B = np.array(['b1', 'b2'])
A_C = np.array(['c1', 'c2'])
```

```python
R_A_a1 = np.array([[1, 2], [2, 0]])
R_B_a1 = np.array([[3, 0], [0, 1]])
R_C_a1 = np.array([[1, 3], [2, 2]])
```

```python
G_2_4_a1 = { 'B' : (R_B_a1, A_B), 'C' : (R_C_a1, A_C) }
```

```python
nash(G_2_4_a1)
```

```python
FP(G_2_4_a1, niter = 100000)
```

```python
print('NE -> B: ({:.3f}, {:.3f}) C: ({:.3f}, {:.3f})'.format(1/2, 1/2, 1/4, 3/4)) 
```

<p>Equilibrios de Nash cuando $A$ juega $a_1$
\begin{align*}
\pi_B &= \bigg( \frac{1}{2}, \frac{1}{2}  \bigg)    &
\pi_C &= \bigg( \frac{1}{4}, \frac{3}{4}  \bigg)    
\end{align*}

```python
NE_B_a1 = np.array([1/2, 1/2])
NE_C_a1 = np.array([1/4, 3/4])
```

<p>Valor para $A$

```python
V_A_a1 = np.sum(np.matmul(R_A_a1, NE_C_a1) * NE_B_a1)
V_A_a1
```

<p>Valor para $B$

```python
V_B_a1 = np.sum(np.matmul(R_B_a1, NE_C_a1) * NE_B_a1)
V_B_a1
```

<p>Valor para $C$

```python
V_C_a1 = np.sum(np.matmul(R_C_a1, NE_B_a1) * NE_C_a1)
V_C_a1
```

```python
V_a1 = np.array([V_A_a1, V_B_a1, V_C_a1])
V_a1
```

```python
R_A_a2 = np.array([[2, 0], [1, 2]])
R_B_a2 = np.array([[0, 1], [2, 0]])
R_C_a2 = np.array([[3, 1], [2, 3]])
```

```python
G_2_4_a2 = { 'B' : (R_B_a2, A_B), 'C' : (R_C_a2, A_C) }
```

```python
nash(G_2_4_a2)
```

```python
FP(G_2_4_a2, niter = 100000)
```

```python
print('NE -> B: ({:.3f}, {:.3f}) C: ({:.3f}, {:.3f})'.format(2/3, 1/3, 1/3, 2/3)) 
```

<p>Equilibrios de Nash cuando $A$ juega $a_2$
\begin{align*}
\pi_B &= \bigg( \frac{2}{3}, \frac{1}{3}  \bigg)    &
\pi_C &= \bigg( \frac{1}{3}, \frac{2}{3}  \bigg)    
\end{align*}

```python
NE_B_a2 = np.array([2/3, 1/3])
NE_C_a2 = np.array([1/3, 2/3])
```

<p>Valor para $A$

```python
V_A_a2 = np.sum(np.matmul(R_A_a2, NE_C_a2) * NE_B_a2)
V_A_a2
```

<p>Valor para $B$

```python
V_B_a2 = np.sum(np.matmul(R_B_a2, NE_C_a2) * NE_B_a2)
V_B_a2
```

<p>Valor para $C$

```python
V_C_a2 = np.sum(np.matmul(R_C_a2, NE_B_a2) * NE_C_a2)
V_C_a2
```

```python
V_a2 = np.array([V_A_a2, V_B_a2, V_C_a2])
V_a2
```
