```python
from games.rps import RPS
from agents.fictitiousplay import FictitiousPlay
from agents.random_agent import RandomAgent
from agents.regretmatching import RegretMatching

N_EPISODES = 50000

game = RPS()
game.reset()

agents = {
    'agent_0': FictitiousPlay(game=game, agent='agent_0'),
    'agent_1': RegretMatching(game=game, agent='agent_1'),
}

rewards_acum = {'agent_0': 0, 'agent_1': 0}

for i in range(N_EPISODES):
    actions = {a: agents[a].action() for a in game.agents}
    game.step(actions)
    for a in game.agents:
        rewards_acum[a] += game.reward(a)

print(f"Rewards acumulados tras {N_EPISODES} episodios:")
for a in game.agents:
    print(f"  {a}: {rewards_acum[a]:.1f}")

print()
print("Políticas aprendidas:")
for a in game.agents:
    print(f"  {a}: {agents[a].policy()}")
```
